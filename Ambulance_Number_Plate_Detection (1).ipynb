```python
import os
import glob

base = "/content/drive/MyDrive/Ambulance_Project/Final_Dataset"

for split in ["train", "valid", "test"]:
    label_dir = os.path.join(base, split, "labels")
    files = glob.glob(os.path.join(label_dir, "*.txt"))

    valid_count = 0
    invalid_count = 0

    for file in files:
        with open(file, "r") as f:
            lines = f.readlines()

        for line in lines:
            values = line.strip().split()

            try:
                nums = list(map(float, values))
                assert len(nums) == 5
                assert nums[0] == 0
                assert all(0 <= x <= 1 for x in nums[1:])
                assert nums[3] > 0 and nums[4] > 0
                valid_count += 1
            except:
                invalid_count += 1

    print(f"\n{split.upper()}")
    print("Valid boxes:", valid_count)
    print("Invalid boxes:", invalid_count)
```

```python
import os
import pandas as pd
from ultralytics import YOLO

# Load pretrained model
model = YOLO("/content/drive/MyDrive/Ambulance_Project/best.pt")

dataset_path = "/content/drive/MyDrive/Ambulance_Project/Final_Dataset"

results_path = "/content/drive/MyDrive/Ambulance_Project/Detection_Results"
os.makedirs(results_path, exist_ok=True)

all_results = []

for split in ["train", "valid", "test"]:

    image_folder = os.path.join(dataset_path, split, "images")

    print(f"\nProcessing {split.upper()} images...")

    results = model.predict(
        source=image_folder,
        conf=0.25,
        imgsz=640,
        save=True,
        project=results_path,
        name=split,
        exist_ok=True,
        device=0
    )

    for result in results:
        image_name = os.path.basename(result.path)
        count = len(result.boxes)

        all_results.append({
            "Dataset": split,
            "Image": image_name,
            "Number_of_Plates": count,
            "Detection_Status": "Detected" if count > 0 else "Not Detected"
        })

    print(f"{split.upper()} Completed!")

# Save detection summary
df = pd.DataFrame(all_results)

csv_path = os.path.join(results_path, "detection_summary.csv")
df.to_csv(csv_path, index=False)

print("\nDetection Completed!")
print(df["Detection_Status"].value_counts())
print("Total Images:", len(df))
print("CSV Saved:", csv_path)
```

```python
import os
import pandas as pd
from ultralytics import YOLO

model = YOLO("/content/drive/MyDrive/Ambulance_Project/best.pt")

dataset_path = "/content/drive/MyDrive/Ambulance_Project/Final_Dataset"

results_path = "/content/drive/MyDrive/Ambulance_Project/Detection_Results"
os.makedirs(results_path, exist_ok=True)

all_results = []

for split in ["train", "valid", "test"]:

    image_folder = os.path.join(dataset_path, split, "images")

    images = [
        os.path.join(image_folder, f)
        for f in sorted(os.listdir(image_folder))
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    print(f"\nProcessing {split.upper()} : {len(images)} images")

    for start in range(0, len(images), 16):

        batch = images[start:start+16]

        results = model.predict(
            source=batch,
            conf=0.25,
            imgsz=640,
            batch=16,
            device=0,
            save=False,
            verbose=False
        )

        for result in results:
            all_results.append({
                "Dataset": split,
                "Image": os.path.basename(result.path),
                "Number_of_Plates": len(result.boxes),
                "Detection_Status": (
                    "Detected" if len(result.boxes) > 0
                    else "Not Detected"
                )
            })

        print(f"Completed: {min(start+16, len(images))}/{len(images)}")

    # Save progress after each split
    df = pd.DataFrame(all_results)
    df.to_csv(
        os.path.join(results_path, "detection_summary.csv"),
        index=False
    )

print("\nAll Images Processing Completed!")
print(pd.DataFrame(all_results)["Detection_Status"].value_counts())
```

```python
import os
import cv2
import easyocr
import pandas as pd
from ultralytics import YOLO

model = YOLO("/content/drive/MyDrive/Ambulance_Project/best.pt")

reader = easyocr.Reader(['en'], gpu=True)

dataset_path = "/content/drive/MyDrive/Ambulance_Project/Final_Dataset"

ocr_results = []

for split in ["train", "valid", "test"]:

    image_folder = os.path.join(dataset_path, split, "images")

    images = [
        f for f in os.listdir(image_folder)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    print(f"\nProcessing {split.upper()}...")

    for i, image_name in enumerate(sorted(images)):

        image_path = os.path.join(image_folder, image_name)
        image = cv2.imread(image_path)

        results = model.predict(
            image,
            conf=0.25,
            imgsz=640,
            device=0,
            verbose=False
        )

        for box in results[0].boxes:

            x1, y1, x2, y2 = map(
                int, box.xyxy[0].tolist()
            )

            crop = image[y1:y2, x1:x2]

            if crop.size == 0:
                continue

            gray = cv2.cvtColor(
                crop, cv2.COLOR_BGR2GRAY
            )

            texts = reader.readtext(gray)

            plate_text = " ".join(
                [text[1] for text in texts]
            )

            ocr_results.append({
                "Dataset": split,
                "Image": image_name,
                "Number_Plate_Text": plate_text
            })

        if (i + 1) % 25 == 0:
            print(f"Completed: {i+1}/{len(images)}")

    pd.DataFrame(ocr_results).to_csv(
        "/content/drive/MyDrive/Ambulance_Project/ocr_results.csv",
        index=False
    )

print("\nOCR Processing Completed!")
print("Total OCR Records:", len(ocr_results))
```

```python

import pandas as pd
import re

df = pd.read_csv(
    "/content/drive/MyDrive/Ambulance_Project/ocr_results.csv"
)

def clean_plate(text):
    if pd.isna(text):
        return ""

    text = re.sub(r"[^A-Za-z0-9]", "", str(text)).upper()
    return text

df["Cleaned_Text"] = df["Number_Plate_Text"].apply(clean_plate)

# Indian vehicle registration number pattern
pattern = r"^[A-Z]{2}[0-9]{1,2}[A-Z]{1,3}[0-9]{1,4}$"

df["Possible_Number_Plate"] = df["Cleaned_Text"].apply(
    lambda x: "Yes" if re.match(pattern, x) else "No"
)

print("Possible Number Plates:")
display(df[df["Possible_Number_Plate"] == "Yes"].head(30))

print("\nTotal Possible Number Plates:",
      (df["Possible_Number_Plate"] == "Yes").sum())
```

```python

import pandas as pd
import os

base_path = "/content/drive/MyDrive/Ambulance_Project"

df = pd.read_csv(os.path.join(base_path, "ocr_results.csv"))

def clean_plate(text):
    if pd.isna(text):
        return ""
    import re
    return re.sub(r"[^A-Za-z0-9]", "", str(text)).upper()

df["Cleaned_Text"] = df["Number_Plate_Text"].apply(clean_plate)

import re
pattern = r"^[A-Z]{2}[0-9]{1,2}[A-Z]{1,3}[0-9]{1,4}$"

df["Possible_Number_Plate"] = df["Cleaned_Text"].apply(
    lambda x: "Yes" if re.match(pattern, x) else "No"
)

possible = df[df["Possible_Number_Plate"] == "Yes"].copy()

summary = df.groupby("Dataset").agg(
    Total_Records=("Image", "count"),
    Possible_Plates=("Possible_Number_Plate",
                     lambda x: (x == "Yes").sum())
).reset_index()

excel_path = os.path.join(base_path, "Final_OCR_Results.xlsx")

with pd.ExcelWriter(excel_path, engine="openpyxl") as writer:
    df.to_excel(writer, sheet_name="All_OCR_Results", index=False)
    possible.to_excel(writer, sheet_name="Possible_Plates", index=False)
    summary.to_excel(writer, sheet_name="OCR_Summary", index=False)

print("Excel file saved successfully!")
print(excel_path)
```
